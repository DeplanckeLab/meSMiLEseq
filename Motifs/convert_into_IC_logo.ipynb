```python
import pandas as pd
import numpy as np
import sys
import random
import logomaker
import os
import matplotlib.pyplot as plt
import glob
```

```python
this_path = os.getcwd()
results_path = '/home/gralak/updepla/users/gralak/SmileSeq_paper/meSMiLEseq_separated_analysis/psam_used_in_pub/IC_logo'
sys.path.append('/data/gralak/meSMiLEseq_github/meSMiLEseq/separate_analysis')
import functions
```

```python
files = glob.glob('/home/gralak/updepla/users/gralak/SmileSeq_paper/meSMiLEseq_separated_analysis/psam_used_in_pub/*_bindingmode_1_ppm.csv')
```

```python
#all logos
for file in files:
    name = file.split('/')[-1][:-4]
    ppm = pd.read_csv(file, index_col=0)

    # First convert into pwm
    pwm = functions.create_PWM(ppm)

    # Calculate information content with background of 0.25
    im = functions.convert_PPM_to_IM(ppm, pwm)
    
    
    fig, ax = plt.subplots(1,1,figsize=[10,2.5])

    logo = logomaker.Logo(im.T.reset_index(drop=True),
                    ax=ax,
                    width=.8,
                    vpad=.05,
                    fade_probabilities=True,
                    stack_order='small_on_top',
                    color_scheme={'A':'#66a61e', 'C':'#7570b3','G':'#ffc809','T':'#d95f02','m':'#a6cee3'}
                    )
    # style using Logo methods
    logo.style_spines(spines=['left', 'right'], visible=False)


    # style using Axes methods
    logo.ax.set_ylabel('information content')
    #logo.ax.set_yticks([0, 1, 2])
    logo.ax.set_xticks(range(len(im.T.reset_index(drop=True))))


    fig.savefig(os.path.join(results_path, f'{name}_ic.pdf'), format='pdf')
    fig.savefig(os.path.join(results_path, f'{name}_ic.png'), format='png', dpi= 600)
    plt.show()
```
