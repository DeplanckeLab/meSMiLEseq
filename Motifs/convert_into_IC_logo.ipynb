```python
import pandas as pd
import sys
import logomaker
import os
import matplotlib.pyplot as plt
import glob
```

#### Disclaimer!

To run this script, first you need to generate position probability matrices, so ppms. Store them in a folder that you pass to this script.

```python
this_path = os.getcwd()
results_path = 'your/results/folder'
sys.path.append('../meSMiLEseq/')
import utils
```

```python
files = glob.glob('/your/ppm/folder/*_bindingmode_1_ppm.csv')
```

```python
#all logos
for file in files:
    name = file.split('/')[-1][:-4]
    ppm = pd.read_csv(file, index_col=0)

    # First convert into pwm
    pwm = utils.create_PWM(ppm)

    # Calculate information content with background of 0.25
    im = utils.convert_PPM_to_IM(ppm, pwm)
    
    
    fig, ax = plt.subplots(1,1,figsize=[10,2.5])

    logo = logomaker.Logo(im.T.reset_index(drop=True),
                    ax=ax,
                    width=.8,
                    vpad=.05,
                    fade_probabilities=False,
                    #stack_order='small_on_top',
                    color_scheme={'A':'#66a61e', 'C':'#7570b3','G':'#ffc809','T':'#d95f02','m':'#a6cee3'}
                    )
    # style using Logo methods
    logo.style_spines(spines=['left', 'right', 'top'], visible=False)


    # style using Axes methods
    logo.ax.set_ylabel('information content')
    logo.ax.set_yticks([0, 1, 2])
    logo.ax.set_xticks(range(len(im.T.reset_index(drop=True))))


    fig.savefig(os.path.join(results_path, f'{name}_ic.pdf'), format='pdf')
    fig.savefig(os.path.join(results_path, f'{name}_ic.png'), format='png', dpi= 600)
    #plt.show()
```
