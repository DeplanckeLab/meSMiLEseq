```python
import pandas as pd
import os
import shutil
```

```python
metafile = pd.read_csv('/data/gralak/meSMiLEseq_github/meSMiLEseq/TF_with_motifs_new.csv')
```

```python
metafile
```

```python
datapath = '/home/gralak/updepla/users/gralak/SmileSeq_paper/meSMiLEseq_separated_analysis/psam/'
outputpath = '/home/gralak/updepla/users/gralak/SmileSeq_paper/meSMiLEseq_separated_analysis/psam_used_in_pub'
```

```python
not_found = []
for i in range(len(metafile)):
    for spec in ['.csv', '_logo.pdf']:
        try:
            a_methylated = os.path.join(
                datapath, 
                metafile.TF[i], 
                metafile.exp[i].replace('exp','SmSAG'),             
                'methylated', 
                f'{metafile.TF[i]}_bm{str(metafile.binding_mode_size[i])}_methylated_bindingmode_1{spec}'
                )
            b_methylated = os.path.join(
                outputpath,
                f'{metafile.TF[i]}_bm{str(metafile.binding_mode_size[i])}_methylated_bindingmode_1{spec}'
                )

            shutil.copy(a_methylated, b_methylated)

            a_unmethylated = os.path.join(
                datapath, 
                metafile.TF[i], 
                metafile.exp[i].replace('exp','SmSAG'), 
                'unmethylated', 
                f'{metafile.TF[i]}_bm{str(metafile.binding_mode_size[i])}_unmethylated_bindingmode_1{spec}'
                )
            b_unmethylated =os.path.join(
                outputpath,
                f'{metafile.TF[i]}_bm{str(metafile.binding_mode_size[i])}_unmethylated_bindingmode_1{spec}'
                )

            shutil.copy(a_unmethylated, b_unmethylated)
        except FileNotFoundError:
            not_found.append(f'{metafile.TF[i]}_{metafile.exp[i]}_bm{str(metafile.binding_mode_size[i])}')

print(not_found)


```
