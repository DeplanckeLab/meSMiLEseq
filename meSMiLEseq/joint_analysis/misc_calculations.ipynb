```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
import shutil
```

```python
path = '/home/gralak/updepla/users/gralak/SmileSeq_paper/meSMiLEseq_joint_analysis/'
experiments = ['SmSAG01', 'SmSAG02', 'SmSAG03', 'SmSAG04', 'SmSAG05', 
               'SmSAG06', 'SmSAG07', 'SmSAG08', 'SmSAG09', 'SmSAG10', 
               'SmSAG11', 'SmSAG12', 'SmSAG13', 'SmSAG14', 'SmSAG15', 
               'SmSAG16', 'SmSAG17', 'SmSAG18', 'SmSAG19', 'SmSAG20', 
               'SmSAG21', 'SmSAG22', 'SmSAG23' 
               ]
```

```python
for experiment in experiments:
    exp_path = path + experiment + '/02_fishers_exact_test/'
    out_path = exp_path + 'significant_kmers/'

    try:
        os.mkdir(out_path)
    except FileExistsError:
        pass

    files = [f for f in os.listdir(exp_path) if f.endswith('pvalues.csv')]

    log_file_path = os.path.join(exp_path, f"{experiment}_significance_log.txt")

    with open(log_file_path, 'w') as log_file:
        log_file.write(f"Significance log for {experiment}\n")
        log_file.write("=====================================\n")

        for file in files:
            result_df = pd.read_csv(exp_path + file)
            threshold = 0.05

            result_df['below_threshold'] = result_df['p_adjust'] < threshold


            significant = result_df[result_df['below_threshold'] == True]

            num_significant = len(significant)
            total = len(result_df)

            percentage = (num_significant / total) * 100 if total > 0 else 0

            
                    
            if len(significant) >= 1:
                new_filename = file.replace('pvalues.csv', 'significant.csv')
                significant.to_csv(out_path + new_filename, index=False)

        
            log_file.write(f"{file} --> {num_significant} significant kmers --> {percentage:.2f}%\n")
```

```python
TFs_with_motifs = pd.read_csv('/data/gralak/meSMiLEseq_github/meSMiLEseq/TF_with_motifs_latest_curation.csv')
```

```python
# Define a bunch of paths, they will all be separated by either TF names or other things
base_path = '/home/gralak/updepla/users/gralak/SmileSeq_paper/'

joint_analysis_path = 'meSMiLEseq_joint_analysis/'
motif_path = '04_ProBound_analysis/psam/'
significant_kmer_path = '02_fishers_exact_test/significant_kmers/'
kmer_ratio_path = '03_kmer_ratios/with_slopes/'

separated_analysis_path = 'meSMiLEseq_separated_analysis/psam/'
methylated_path = 'methylated/'
unmethylated_path = 'unmethylated/'

output_base = os.path.join(base_path, 'meSMiLEseq_motifs_and_scatterplots_for_publication/')
out_joint_matrix = 'joint/matrices/'
out_joint_plots = 'joint/plots/'
out_sep_matrix = 'separated/matrices/'
out_sep_plots = 'separated/plots/'
```

```python
for TF, df in TFs_with_motifs.groupby('TF'):
    TFname = TF
    SMSexp = metadata_dict[df.exp.values[0]]
    bs = df.binding_mode_size.values[0] #it's a integer!
    scatter = df.kmer_scatter.values[0]

    #Create output_paths
    out_TF_jm = os.path.join(output_base, TFname, out_joint_matrix)
    out_TF_jp = os.path.join(output_base, TFname, out_joint_plots)
    out_TF_sm = os.path.join(output_base, TFname, out_sep_matrix)
    out_TF_sp = os.path.join(output_base, TFname, out_sep_plots)

    os.makedirs(out_TF_jm, exist_ok=True)
    os.makedirs(out_TF_jp, exist_ok=True)
    os.makedirs(out_TF_sm, exist_ok=True)
    os.makedirs(out_TF_sp, exist_ok=True)
    
    
    #First copy the joint
    src_joint = os.path.join(base_path, joint_analysis_path, SMSexp)
    scr_joint_motif_matrix = os.path.join(src_joint, motif_path, TFname, f'binding_mode_size_{str(bs)}/', f'{TFname}_bindingmode_1.csv')
    scr_joint_motif_pdf = os.path.join(src_joint, motif_path, TFname, f'binding_mode_size_{str(bs)}/', f'{TFname}_bindingmode_1_logo.pdf')
    src_joint_pval = os.path.join(src_joint, significant_kmer_path, f'{TFname}_{scatter}_significant.csv')
    src_joint_ratio = os.path.join(src_joint, kmer_ratio_path, f'{TFname}_{scatter}_ratios.csv')

    shutil.copy(scr_joint_motif_matrix, out_TF_jm)
    shutil.copy(scr_joint_motif_pdf, out_TF_jp)
    shutil.copy(src_joint_ratio, out_TF_jm)
    try:
        shutil.copy(src_joint_pval, out_TF_jm)
    except FileNotFoundError:
        pass

    #Second copy the separated
    src_sep = os.path.join(base_path, separated_analysis_path, TFname)
    scr_sep_m_motif_matrix = os.path.join(src_sep, SMSexp, methylated_path, f'{TFname}_bm{str(bs)}_methylated_bindingmode_1.csv')
    scr_sep_nm_motif_matrix = os.path.join(src_sep, SMSexp, unmethylated_path, f'{TFname}_bm{str(bs)}_unmethylated_bindingmode_1.csv')
    scr_sep_m_motif_pdf = os.path.join(src_sep, SMSexp, methylated_path, f'{TFname}_bm{str(bs)}_methylated_bindingmode_1_logo.pdf')
    scr_sep_nm_motif_pdf = os.path.join(src_sep, SMSexp, unmethylated_path, f'{TFname}_bm{str(bs)}_unmethylated_bindingmode_1_logo.pdf')

    shutil.copy(scr_sep_m_motif_matrix, out_TF_sm)
    shutil.copy(scr_sep_nm_motif_matrix, out_TF_sm)
    shutil.copy(scr_sep_m_motif_pdf, out_TF_sp)
    shutil.copy(scr_sep_nm_motif_pdf, out_TF_sp)


    
    
```
