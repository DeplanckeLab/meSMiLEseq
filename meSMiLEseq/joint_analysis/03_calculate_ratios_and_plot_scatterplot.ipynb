```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
import multiprocessing
import sys



print('####################################################################')
print('#### Calculate ratios for scatterplots ©Antoni Gralak_01.07.2025 ###')
print('####################################################################')
print('Setting env...')

this_path = os.getcwd()
sys.path.append(os.path.join(this_path, '..'))

import utils

num_cores = 1
#num_cores = 23
experiment_names = ['exp10']
                #['exp1', 'exp2', 'exp3', 'exp4', 'exp5',
                #'exp6', 'exp7', 'exp8', 'exp9', 'exp10',
                #'exp11', 'exp12', 'exp13', 'exp14', 'exp15',
                #'exp16', 'exp17', 'exp18', 'exp19', 'exp20',
                #'exp21', 'exp22', 'exp23']
kmers = [6] #[6,7,8,9]
to_be_analyzed = ['ZNF891_FL']

```

```python
def process_experiment(experiment_name):
    
    data_path = '../output_joint_analysis/'
    
    for TF in to_be_analyzed:
        for kmer in kmers:
            kmer_path = data_path + f'/01_kmer_analysis/{experiment_name}_{TF}_{kmer}mer_enrichment.csv'
    
            save_path = data_path + '/03_calculate_ratios/'

            try:
                os.makedirs(save_path)
            except FileExistsError:
                pass



            print(f'Calculating ratios for {TF} ({experiment_name}, {kmer}) with scatterplot..\n')
            print('#############')

            kmer_df = pd.read_csv(kmer_path)
            final_df = utils.calculate_ratios(kmer_df)


            final_df.to_csv(save_path + f'{experiment_name}_{TF}_{kmer}mer_ratio.csv', index=False)

                
            lightblack = '#333333'
            darkred = '#FF0000'

            x = final_df['eluted_methl']
            y = final_df['eluted_nonmethl']

            fig, ax = plt.subplots(1, 1, figsize=(5.5, 5.5))
            colors = {True: darkred, False: lightblack}

            ax.scatter(x=x, y=y, alpha=0.6, facecolors='none',
                        edgecolors=final_df['CpG'].map(colors), rasterized=True)
            ax.grid(visible=False)

            ax.spines['top'].set_visible(False)
            ax.spines['right'].set_visible(False)

            lower_limit = None
            max_val = max(x.max(), y.max())
            ax.set_xlim(lower_limit, max_val * 1.05)
            ax.set_ylim(lower_limit, max_val * 1.05)
            ax.set_aspect('equal', adjustable='box')

            ax.set_xlabel('methylated DNA', fontfamily='sans-serif', fontsize=10, fontstyle='italic')
            ax.set_ylabel('unmethylated DNA', fontfamily='sans-serif', fontsize=10, fontstyle='italic')
            ax.set_title(f"{TF.replace('_', ' ')}, {kmer} enrichment, normalized by input", fontsize=10)
            plt.savefig(save_path + f'{experiment_name}_{TF}_{kmer}mer_scatterplot.pdf', dpi=400, bbox_inches='tight')
            plt.savefig(save_path + f'{experiment_name}_{TF}_{kmer}mer_scatterplot.png', dpi=400, bbox_inches='tight')
            plt.close()

            print('Done!')
```

```python
if __name__ == '__main__':

    with multiprocessing.Pool(num_cores) as pool:
        pool.map(process_experiment, experiment_names)
```
