# Exploring prettiness

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import ks_2samp
import pandas as pd

# Example: you already have hbonds_um_usf and hbonds_m_usf arrays
time_ps_1, hbonds_1 = hbonds_um_usf[:,0], hbonds_um_usf[:,1]
time_ps_2, hbonds_2 = hbonds_m_usf[:,0], hbonds_m_usf[:,1]

# Convert to ns
time_ns_1 = time_ps_1 / 1000
time_ns_2 = time_ps_2 / 1000

# Optional: smooth with rolling mean (window=1000 points ~ adjust as needed)
#hbonds_1_smooth = pd.Series(hbonds_1).rolling(window=1000, min_periods=1).mean()
#hbonds_2_smooth = pd.Series(hbonds_2).rolling(window=1000, min_periods=1).mean()

# Plot
plt.figure(figsize=(10,6))

plt.plot(time_ns_1, hbonds_1, lw=1.5, color="tab:blue", label="unmethylated DNA")
plt.plot(time_ns_2, hbonds_2, lw=1.5, color="black", label="methylated DNA")

# Style
plt.xlabel("Time (ns)", fontsize=14)
plt.ylabel("Number of H-bonds", fontsize=14)
plt.title("H-bonds in USF3-DNA complex", fontsize=16, weight="bold")

plt.xticks(np.arange(0, 801, 100), fontsize=12)
plt.yticks(fontsize=12)

plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(frameon=False, fontsize=12)

# Remove top/right spines for a clean look
for spine in ["top","right"]:
    plt.gca().spines[spine].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
# KS test on raw hbonds distributions
ks_stat, p_value = ks_2samp(hbonds_1, hbonds_2)

print(f"KS statistic = {ks_stat:.4f}, p-value = {p_value:.4e}")
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


# --- Example: you already have these arrays ---
# hbonds_um_usf = np.array([...])
# hbonds_other = np.array([...])

time_ps_1, hbonds_1 = hbonds_um_usf[:,0], hbonds_um_usf[:,1]
time_ps_2, hbonds_2 = hbonds_m_usf[:,0], hbonds_m_usf[:,1]

# Convert to ns
time_ns_1 = time_ps_1 / 1000
time_ns_2 = time_ps_2 / 1000

# Optional smoothing (rolling mean with window = 1000 points)
hbonds_1_smooth = pd.Series(hbonds_1).rolling(window=10, min_periods=1).mean()
hbonds_2_smooth = pd.Series(hbonds_2).rolling(window=10, min_periods=1).mean()

# Put data into a single DataFrame for Seaborn
df = pd.DataFrame({
    "Time (ns)": np.concatenate([time_ns_1, time_ns_2]),
    "H-bonds": np.concatenate([hbonds_1_smooth, hbonds_2_smooth]),
    "Condition": (["unmethylated"] * len(time_ns_1)) + (["methylated"] * len(time_ns_2))
})

# --- Plot with Seaborn ---
sns.set_theme(style="whitegrid")  # nice defaults

plt.figure(figsize=(10,6))
sns.lineplot(
    data=df,
    x="Time (ns)", y="H-bonds", hue="Condition",
    palette=["tab:blue", "black"],
    linewidth=1.5
)

plt.title("UF3", fontsize=14)
plt.xlabel("Time (ns)", fontsize=14)
plt.ylabel("Number of H-bonds", fontsize=14)
plt.xticks(np.arange(0, 801, 100), fontsize=12)
plt.yticks(fontsize=12)

plt.legend(title="", frameon=False, fontsize=12)
plt.tight_layout()
plt.savefig('/home/gralak/updepla/users/gralak/meSMS_20250812_Revisions/Parths_analysis/USF3_joint_plot_averaged_by10.svg', format="svg", dpi=400)
plt.show()


```
